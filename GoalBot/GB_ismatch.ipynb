```python
# 2
# Optional: Ensure consistent types (if needed)
df1['salesDetails_id'] = df1['salesDetails_id'].astype(int)
df2['salesDetails_id'] = df2['salesDetails_id'].astype(int)

# Merge on salesDetails_id (df1 priority)
merged = df1.merge(df2, on='salesDetails_id', suffixes=('_df1', '_df2'))

# Filter where invoiceDate matches
matched = merged[merged['invoiceDate_df1'] == merged['invoiceDate_df2']]

# Then compare other fields one by one, treating NaNs as equal using fillna
matched = matched[
    (matched['stockistId_df1'].fillna(-1) == matched['stockistId_df2'].fillna(-1)) &
    (matched['chemistId_df1'].fillna(-1) == matched['chemistId_df2'].fillna(-1)) &
    (matched['productId_df1'] == matched['productId_df2']) &
    (matched['qty_df1'] == matched['qty_df2']) &
    (matched['rate_df1'] == matched['rate_df2'])
]

# Output matching salesDetails_id
matching_ids = matched['salesDetails_id'].unique().tolist()
print(matching_ids)
```
