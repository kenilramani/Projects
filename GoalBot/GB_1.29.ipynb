```python
product_sales_frequency = df['productId'].value_counts()
# product_sales_frequency

plt.figure(figsize=(15, 6)) 

sns.barplot(
    x=product_sales_frequency.index,
    y=product_sales_frequency.values,
    palette='viridis' 
)

# Add title and labels
plt.title('product sale frequency', fontsize=16)
plt.xlabel('productId', fontsize=12)
plt.ylabel('times product is sold', fontsize=12)

plt.xticks(rotation=90, ha='right', fontsize=7)

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()

plt.show()

# # --- Optional: Further analysis or insights ---
# print("--- Analysis Insights ---")
# if not product_sales_frequency.empty:
#     most_sold_product = product_sales_frequency.index[0]
#     most_sold_count = product_sales_frequency.values[0]
#     print(f"The most frequently sold product is '{most_sold_product}', sold {most_sold_count} times.")
# else:
#     print("No product sales data to analyze.")

```
